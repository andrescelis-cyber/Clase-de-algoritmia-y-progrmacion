```python

import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# 1. Parámetros del problema del circuito RL
V = 100.0   # Voltios
L = 2.0     # Henrios
t = 10.0    # Segundos
i_target = 10.0 # Amperios

# Tolerancia en error relativo (0.05% = 0.0005)
e = 0.05 / 100

# 2. Definición de la función f(R)
def f(R):
    return (V / R) * (1 - np.exp(-(R / L) * t)) - i_target

# 3. Ingresar el intervalo inicial de búsqueda (por ejemplo: [1, 15] o [1, 50])
a = float(input('Intervalo inferior (a): '))
b = float(input('Intervalo superior (b): '))

# Evaluaciones iniciales
fi = f(a)
fs = f(b)

# Asegurar que la función cambia de signo en el intervalo [a, b]
while np.sign(fi) == np.sign(fs):
    a -= 1
    if a <= 0:
        a = 0.1  # Evitar R <= 0 por división entre cero
    b += 1
    fi = f(a)
    fs = f(b)

# 4. Configuración y despliegue de la gráfica inicial
x_vals = np.linspace(a, b, 400)
y_vals = f(x_vals)

plt.figure(figsize=(8, 5))
plt.plot(x_vals, y_vals, linewidth=2, label='f(R)')
plt.axhline(0, color='red', linestyle='--', linewidth=1) # Línea en y=0
plt.xlabel('Resistencia R (Ohmios)')
plt.ylabel('f(R)')
plt.title('Método de Bisección - Circuito RL')
plt.grid(True)

# 5. Algoritmo del Método de Bisección
E = 1.0  # Error inicial
A = []   # Lista para almacenar los resultados por iteración
i = 1    # Contador de iteraciones
c_old = a

while E > e:
    c = 0.5 * (a + b)
    fc = f(c)

    # Cálculo del error relativo
    if i > 1:
        E = abs((c - c_old) / c)
    else:
        E = 1.0

    c_old = c

    # Almacenar datos para la tabla [n, a, b, c, f(c), error]
    A.append([i, a, b, c, fc, E])

    # Verificación de cambio de signo para actualizar límites
    if fi * fc < 0:
        b = c
        fs = fc
    else:
        a = c
        fi = fc

    i += 1

# 6. Mostrar Resultados
raiz_aprox = A[-1][3]
print(f"\nLa raíz aproximada es R* = {raiz_aprox:.6f} Ohmios")

# Crear y mostrar la tabla de resultados
T = pd.DataFrame(A, columns=['n', 'a', 'b', 'c', 'f(c)', 'error'])
print("\nTabla de Iteraciones:")
display(T)

# Graficar el punto de la raíz sobre la gráfica generada
plt.plot(raiz_aprox, f(raiz_aprox), 'rd', markersize=8, label=f'Raíz aproximada ({raiz_aprox:.4f})')
plt.legend()
plt.show()
```
